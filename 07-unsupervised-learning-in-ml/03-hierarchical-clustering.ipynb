```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dataset = pd.read_csv("../assets/csv-files/iris-flower.csv")
dataset.head(3)
```

```python
# We need a unlabeled dataset
# Converting the dataset to unlabeled
dataset.drop(columns="Species", inplace=True)
dataset.head(3)
```

```python
# Visualize data
sns.pairplot(dataset)
plt.show()
```

```python
# Create a dendrogram
import scipy.cluster.hierarchy as sc
```

```python
plt.figure(figsize=(50, 25))
sc.dendrogram(sc.linkage(dataset, method="single", metric="euclidean"))
plt.show()
```

```python
# Make the clusters
from sklearn.cluster import AgglomerativeClustering
```

```python
ac = AgglomerativeClustering(n_clusters=2, linkage="single")
ac.fit_predict(dataset)
```

```python
# Save the predicted cluster value to dataset
dataset["Predict"] = ac.fit_predict(dataset)
dataset.head(3)
```

```python
# Visualize the clusters
sns.pairplot(dataset, hue="Predict")
plt.show()
```
