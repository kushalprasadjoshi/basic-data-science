```python
import pandas as pd
```

```python
dataset = pd.read_csv("../assets/csv-files/groceries.csv")
dataset.head(3)
```

```python
# Check for null values
dataset.isnull().sum()
```

```python
# Create a list over lists removing the null values
market = []
for i in range(0, dataset.shape[0]): # Till all rows finish
    customer = []
    for j in dataset.columns: # Till all columns finish
        if type(dataset[j][i]) is str:
            customer.append(dataset[j][i])
    market.append(customer)
    
# Final list
market
```

```python
# Indentify most frequenty bought products
import collections
```

```python
# Convert the 2D list to 1D list
l = []
for i in market:
    for j in i:
        l.append(j)
# Display list
l
```

```python
# Count total number of repetition
collections.Counter(l) # Takes 1D list
```

```python
# Create a dataset for observation
item_count = collections.Counter(l)
df = {"Item Name" : item_count.keys(), "Repetition": item_count.values()}
dataframe = pd.DataFrame(df).sort_values(by=["Repetition"], ascending=False)
dataframe
```

```python
# Set max number of rows to display
pd.set_option("display.max_rows", 200)
dataframe
```

```python
# Create a confusion matrix for Apriori Algorithm
from mlxtend.preprocessing import TransactionEncoder
```

```python
tr = TransactionEncoder()
tr.fit(market)
```

```python
tr.transform(market)
```

```python
# Create a dataframe of confusion matrix
dframe = pd.DataFrame(tr.transform(market), columns=tr.columns_)
dframe
```

```python
# Use Apriori Algorithm
from mlxtend.frequent_patterns import apriori
```

```python
apriori(dframe, min_support=0.05, use_colnames=True, max_len=3).sort_values(by="support", ascending=False)
```
