```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

#### Classification Analysis

```python
# Create a dataset using make_moons
from sklearn.datasets import make_moons
```

```python
# Get data for dataset
x, y = make_moons(n_samples=1000, noise=0.2)
# Here x is inpendent and y is dependent data
```

```python
# Create a dataframe using x and y for tablular visualization
df = {"x1": x[:, 0], "x2": x[:, 1], "y": y}
dataset = pd.DataFrame(df)
dataset.head(3)
```

```python
# Visualize data
sns.scatterplot(x="x1", y="x2", data=dataset, hue="y")
plt.show()
```

```python
# Train test split
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Classification analysis using Bagging Meta-Estimator technique
from sklearn.ensemble import BaggingClassifier
```

```python
# We need an estimator for  BaggingClassifier
from sklearn.svm import SVC
```

```python
# Check for the score of our estimator
svc = SVC()
svc.fit(x_train, y_train)
svc.score(x_train, y_train), svc.score(x_test, y_test)
```

```python
# Create the BaggingClassifier model
bc = BaggingClassifier(estimator=SVC(), n_estimators=50)
bc.fit(x_train, y_train)
```

```python
bc.score(x_train, y_train), bc.score(x_test, y_test)
```

```python
# Classification analysis using Random Forest technique
from sklearn.ensemble import RandomForestClassifier
```

```python
rfc = RandomForestClassifier(n_estimators=50)
rfc.fit(x_train, y_train)
```

```python
rfc.score(x_train, y_train), rfc.score(x_test, y_test)
```

#### Regression Analysis

```python
dataset = pd.read_csv("../assets/csv-files/maths.csv")
dataset.head(3)
```

```python
# Check for null values
dataset.isnull().sum()
```

```python
# Remove all the null values
dataset.dropna(inplace=True)
```

```python
# Select dependent and indpendent data
x = dataset[["Number3"]]
y = dataset["Total Sum"]
```

```python
# Train test split
from sklearn.model_selection import train_test_split
```

```python
x_train, x_test, y_train, y_test = train_test_split(x, y, test_size=0.2, random_state=42)
```

```python
# Regression analysis using Bagging Meta-Estimator technique
from sklearn.ensemble import BaggingRegressor
```

```python
# We need an estimator for  BaggingRegressor
from sklearn.linear_model import LinearRegression
```

```python
# Check for the score of our estimator
lr = LinearRegression()
lr.fit(x_train, y_train)
lr.score(x_train, y_train), lr.score(x_test, y_test)
```

```python
# Create the BaggingRegressor model
br = BaggingRegressor(estimator=LinearRegression(), n_estimators=50)
br.fit(x_train, y_train)
```

```python
br.score(x_train, y_train), br.score(x_test, y_test)
```

```python
# Regression analysis using Random Forest technique
from sklearn.ensemble import RandomForestRegressor
```

```python
rfr = RandomForestRegressor(n_estimators=50)
rfr.fit(x_train, y_train)
```

```python
rfr.score(x_train, y_train), rfr.score(x_test, y_test)
```
